```python
import zipfile
import pandas as pd

# Load validation dataset
with zipfile.ZipFile("../data/raw/ddxplus/release_validate_patients.zip") as z:
    validate_file = z.namelist()[0]
    validate_df = pd.read_csv(z.open(validate_file))

# Load test dataset
with zipfile.ZipFile("../data/raw/ddxplus/release_test_patients.zip") as z:
    test_file = z.namelist()[0]
    test_df = pd.read_csv(z.open(test_file))

print("Training:", train_df.shape)
print("Validation:", validate_df.shape)
print("Test:", test_df.shape)
```

```python
# Extract unique patient input patterns

train_patterns = get_input_patterns(train_df)
validate_patterns = get_input_patterns(validate_df)
test_patterns = get_input_patterns(test_df)

print("Unique training patterns:", len(train_patterns))
print("Unique validation patterns:", len(validate_patterns))
print("Unique test patterns:", len(test_patterns))

print("\nTrain-Validation overlap:",
      len(train_patterns & validate_patterns))

print("Train-Test overlap:",
      len(train_patterns & test_patterns))

print("Validation-Test overlap:",
      len(validate_patterns & test_patterns))
```

```python
# Input features used to identify an identical patient pattern
input_columns = ["AGE", "SEX", "EVIDENCES"]

# Keep one record per input pattern, along with its actual disease label
train_unique = train_df[
    input_columns + ["PATHOLOGY"]
].drop_duplicates(subset=input_columns)

validate_unique = validate_df[
    input_columns + ["PATHOLOGY"]
].drop_duplicates(subset=input_columns)

test_unique = test_df[
    input_columns + ["PATHOLOGY"]
].drop_duplicates(subset=input_columns)


def check_label_conflicts(df1, df2, split_name):
    merged = df1.merge(
        df2,
        on=input_columns,
        suffixes=("_1", "_2")
    )

    conflicts = merged[
        merged["PATHOLOGY_1"] != merged["PATHOLOGY_2"]
    ]

    print(f"\n{split_name}")
    print("Overlapping patterns:", len(merged))
    print("Conflicting disease labels:", len(conflicts))


check_label_conflicts(
    train_unique, validate_unique, "Train vs Validation"
)

check_label_conflicts(
    train_unique, test_unique, "Train vs Test"
)

check_label_conflicts(
    validate_unique, test_unique, "Validation vs Test"
)
```

```python

from collections import Counter

evidence_counter = Counter()
selected_value_counter = Counter()

for evidence_list in train_df["EVIDENCES"]:

    items = ast.literal_eval(evidence_list)

    for item in items:

        if "_@_" in item:
            evidence_id, selected_value = item.split("_@_", 1)

            evidence_counter[evidence_id] += 1
            selected_value_counter[(evidence_id, selected_value)] += 1

        else:
            evidence_counter[item] += 1

# Evidence frequency summary
evidence_frequency_df = pd.DataFrame(
    evidence_counter.items(),
    columns=["Evidence_ID", "Frequency"]
)

evidence_frequency_df = evidence_frequency_df.sort_values(
    "Frequency",
    ascending=False
)

print("Total unique evidence IDs:", len(evidence_counter))

print("\nMost frequent evidence IDs:")
display(evidence_frequency_df.head(15))

print("\nTotal unique evidence-value combinations:",
      len(selected_value_counter))
```

```python

# Count unique selected values for each evidence ID

unique_values_per_evidence = {}

for evidence_id, selected_value in selected_value_counter:

    unique_values_per_evidence.setdefault(evidence_id, set())
    unique_values_per_evidence[evidence_id].add(selected_value)

# Create summary DataFrame
value_summary = pd.DataFrame([
    {
        "Evidence_ID": evidence_id,
        "Unique_Selected_Values": len(values),
        "Sample_Values": list(values)[:5]
    }
    for evidence_id, values in unique_values_per_evidence.items()
])

value_summary = value_summary.sort_values(
    "Unique_Selected_Values",
    ascending=False
)

print("Evidence IDs having selected values:",
      len(value_summary))

display(value_summary)
```

```python
import ast
import numpy as np
from scipy.sparse import csr_matrix

def convert_to_sparse(df, feature_to_index, feature_names, chunk_size=25000):

    matrices = []

    for start in range(0, len(df), chunk_size):

        chunk = df.iloc[start:start + chunk_size]

        rows = []
        cols = []
        values = []

        for row_index, (age, sex, evidence_string) in enumerate(
            chunk[["AGE", "SEX", "EVIDENCES"]].itertuples(
                index=False, name=None
            )
        ):

            # Add age
            row_features = {
                feature_to_index["AGE"]: age
            }

            # Add sex
            sex_feature = "SEX_" + sex
            row_features[feature_to_index[sex_feature]] = 1

            # Parse evidence list
            items = ast.literal_eval(evidence_string)

            for item in items:

                if "_@_" in item:
                    evidence_id, selected_value = item.split("_@_", 1)
                    feature_name = f"{evidence_id}__{selected_value}"
                else:
                    feature_name = item

                if feature_name not in feature_to_index:
                    raise ValueError(f"Unexpected feature: {feature_name}")

                row_features[feature_to_index[feature_name]] = 1

            # Store row features
            for col_index, value in row_features.items():
                rows.append(row_index)
                cols.append(col_index)
                values.append(value)

        chunk_matrix = csr_matrix(
            (values, (rows, cols)),
            shape=(len(chunk), len(feature_names)),
            dtype=np.float32
        )

        matrices.append(chunk_matrix)

        print(f"Processed {min(start + chunk_size, len(df)):,} / {len(df):,}")

    from scipy.sparse import vstack

    return vstack(matrices, format="csr")
```
